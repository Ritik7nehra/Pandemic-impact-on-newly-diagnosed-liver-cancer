# COVID-19 Liver Cancer EDA

Reproducible exploratory analysis for the dashboard project.

```
import pandas as pd
import plotly.express as px

df = pd.read_csv('../data/covid_liver_clean.csv')
df.head()
```

## Cohort counts

```
df.groupby(['Cohort', 'Cancer_Type_Simple']).size().reset_index(name='cases')
```

```
px.bar(df.groupby(['Cohort', 'Cancer_Type_Simple']).size().reset_index(name='cases'), x='Cohort', y='cases', color='Cancer_Type_Simple', barmode='group', title='Case mix by cohort')
```

## Presentation route

```
presentation = df.groupby(['Cohort', 'Mode_Presentation']).size().reset_index(name='cases')
presentation['percent'] = presentation['cases'] / presentation.groupby('Cohort')['cases'].transform('sum') * 100
presentation
```

```
px.bar(presentation, x='Cohort', y='percent', color='Mode_Presentation', title='Presentation route share')
```

## Survival from MDM

```
df.groupby('Cohort')['Survival_fromMDM'].describe()
```

```
px.box(df, x='Cohort', y='Survival_fromMDM', color='Cohort', points='outliers', title='Survival from MDM')
```

## Missingness

```
missing = (df.isna().mean() * 100).sort_values(ascending=False).reset_index()
missing.columns = ['column', 'missing_percent']
missing.head(15)
```
